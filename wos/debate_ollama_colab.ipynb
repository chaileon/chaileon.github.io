```python
!apt-get install -y pciutils zstd

# 1. Install Ollama
!curl -fsSL https://ollama.com/install.sh | sh

# 2. Start the Ollama server in the background
import subprocess, time
subprocess.Popen(["ollama", "serve"])
time.sleep(5)  # give it a moment to start

```

```python
# 3. Pull the gemma4 model (large download — may take several minutes)
# Pull the model (non-blocking after completion)
subprocess.Popen(["ollama", "run", "gemma4:e4b"])

!ps aux | grep ollama


def check_model_loaded(model_name):
    result = subprocess.run(["ollama", "ps"], capture_output=True, text=True)
    return model_name in result.stdout

while not check_model_loaded("gemma4"):
    print("Model not loaded yet. Waiting...")
    time.sleep(60)
print("model loaded")
```
